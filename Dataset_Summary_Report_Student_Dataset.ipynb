This Google Colab notebook uses **Pandas** to load and explore the Student CSV dataset.

- Inspect the first and last rows
- View random samples

> **Note:** Upload `Day7_Student_Dataset.csv` to the Colab session before running the notebook.


## 1. Import Pandas

```python
import pandas as pd
import numpy as np
```

## 2. Load the CSV Dataset

```python
file_path = "Day7_Student_Dataset.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

## 3. View the First Few Rows

```python
display(df.head())
```

## 4. View the Last Few Rows

```python
display(df.tail())
```

## 5. View Random Samples

```python
display(df.sample(5, random_state=42))
```

## 6. Check Dataset Shape

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Shape:", df.shape)
```

## 7. View Column Names

```python
for i, column in enumerate(df.columns, start=1):
    print(i, ".", column)
```

## 8. View the Dataset Index

```python
print(df.index)
```

## 9. Check Data Types

```python
print(df.dtypes)
```

## 10. Dataset Information Using `info()`

```python
df.info()
```

## 11. Descriptive Statistics Using `describe()`

```python
display(df.describe(include="all").transpose())
```

## 12. Check Missing Values

```python
missing_values = df.isnull().sum()

print(missing_values)
print("\nTotal missing values:", missing_values.sum())
```

## 13. Missing Value Summary

```python
missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (df.isnull().mean().values * 100).round(2)
})

display(missing_summary)
```

## 14. Number of Unique Values

```python
print(df.nunique())
```

## 15. Basic Dataset Observations

```python
print("DATASET SUMMARY")
print("-" * 40)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nAvailable features:")
for column in df.columns:
    print("-", column)

print("\nData types:")
print(df.dtypes)

total_missing = df.isnull().sum().sum()
print("\nTotal missing values:", total_missing)

missing_columns = df.columns[df.isnull().any()].tolist()

if not missing_columns:
    print("Observation: No missing values were found.")
else:
    print("Observation: Missing values were found in:")
    for column in missing_columns:
        print("-", column)
```
