```python
import pandas as pd
import numpy as np
```

```python
file_path = "Day8_Ecommerce_Sales_Dataset.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
display(df.head())
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
# Select one column
display(df["Product"].head())

# Select multiple columns
display(df[["Order_ID", "Product", "Category", "City", "Total_Sales"]].head())
```

```python
# Orders with sales greater than 1000
high_sales = df[df["Total_Sales"] > 1000]

print("Orders with Total_Sales > 1000:")
display(high_sales)
```

```python
# Electronics orders with sales above 500
filtered = df[
    (df["Category"] == "Electronics") &
    (df["Total_Sales"] > 500)
]

display(filtered)
```

```python
# Highest sales first
sorted_sales = df.sort_values("Total_Sales", ascending=False)

display(sorted_sales.head(10))
```

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print("Total Sales      :", total_sales)
print("Average Sales    :", round(average_sales, 2))
print("Highest Sale     :", highest_sales)
print("Lowest Sale      :", lowest_sales)
print("Total Quantity   :", total_quantity)
print("Number of Orders :", number_of_orders)
```

## 9. Sales by Category

```python
category_analysis = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(category_analysis)
```

```python
city_analysis = (
    df.groupby("City")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(city_analysis)
```

```python
product_analysis = (
    df.groupby("Product")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(product_analysis)
```

```python
display(product_analysis.head(5))
```

```python
display(product_analysis.sort_values("Total_Sales").head(5))
```

```python
payment_analysis = (
    df.groupby("Payment_Method")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(payment_analysis)
```

```python
display(
    df.sort_values("Total_Sales", ascending=False)
      .head(10)[[
          "Order_ID", "Product", "Category", "City",
          "Payment_Method", "Quantity", "Total_Sales"
      ]]
)
```

```python
# Orders where more than 3 items were purchased
large_quantity_orders = df[df["Quantity"] > 3]
display(large_quantity_orders)
```

```python
category_city = (
    df.groupby(["Category", "City"])
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(category_city.head(15))
```

```python
payment_category = (
    df.groupby(["Payment_Method", "Category"])
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Number_of_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(payment_category)
```

```python
top_category = category_analysis["Total_Sales"].idxmax()
top_city = city_analysis["Total_Sales"].idxmax()
top_product = product_analysis["Total_Sales"].idxmax()
top_payment = payment_analysis["Total_Sales"].idxmax()

highest_order = df.loc[df["Total_Sales"].idxmax()]

print("1. Top category:", top_category,
      "with sales of", round(category_analysis.loc[top_category, "Total_Sales"], 2))

print("2. Top city:", top_city,
      "with sales of", round(city_analysis.loc[top_city, "Total_Sales"], 2))

print("3. Top product:", top_product,
      "with sales of", round(product_analysis.loc[top_product, "Total_Sales"], 2))

print("4. Leading payment method:", top_payment,
      "with sales of", round(payment_analysis.loc[top_payment, "Total_Sales"], 2))

print("5. Highest-value order:", highest_order["Order_ID"],
      "with sales of", highest_order["Total_Sales"])

print("6. Total quantity sold:", df["Quantity"].sum())
print("7. Number of unique orders:", df["Order_ID"].nunique())
```
